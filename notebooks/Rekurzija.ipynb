```python
def fibon(n):
    print("fibon ", n)
    if n == 1:
        return 1
    if n == 2:
        return 1
    return fibon(n-1) + fibon(n-2)

fibon(7)
```

```python
poteza = 0

def hanoi(od_kod, vmes, do_kam, n, level=0):
    global poteza
    if n == 1:
        poteza += 1
        print(poteza, "   " * level, "Prestavi iz", od_kod, "na", do_kam)
    else:
        hanoi(od_kod, do_kam, vmes, n-1, level+1)
        poteza += 1
        print(poteza, "   " * level, "Prestavi iz", od_kod, "na", do_kam)
        hanoi(vmes, od_kod, do_kam, n-1, level+1)

hanoi("A", "B", "C", 7)
```

```python
def fact(n):
    if n == 0:
        return 1
    return n * fact(n - 1)

fact(5)
```
